```python
import hdb
import pandas as pd
```

```python
def process_data_to_df(data, data_type, items):
    df = pd.DataFrame()
    for name in data_type.dtype.fields:
        field_type = data.dtype.fields[name][0]
        if 0 == field_type.ndim:
            df[name] = data[name]
        elif 1 == field_type.ndim:
            for idx in range(field_type.shape[0]):
                df[name + str(idx)] = data[name][:, idx]

    df['index'] = items['trading_day']
    df['local_time'] = items['local_time']
    df['time_point_seq_no'] = items['time_point_seq_no']
    df['symbol'] = items['symbol']
    df['code'] = df['symbol'].apply(lambda x: x.decode('utf-8'))
    return df
```

```python
data_type = file.data_types[items['type_id'][0]]
data = data_type.items_data(items['data'])
resdf = process_data_to_df(data, data_type, items)
```
